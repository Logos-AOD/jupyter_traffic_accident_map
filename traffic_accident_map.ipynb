Определим, сколько ДТП случалось в каждый день недели. Также покажим число происшествий по месяцам.

```python
# Добавляем столбцы с днем недели и месяцем
days_ru = {
    0: 'Пн', 1: 'Вт', 2: 'Ср', 3: 'Чт',
    4: 'Пт', 5: 'Сб', 6: 'Вс'
}

days_order = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']

traf_accident_kirov['day_of_week'] = pd.Categorical(
    traf_accident_kirov['datetime'].dt.dayofweek.map(days_ru),
    categories=days_order,
    ordered=True
)


months_ru = {
    1: 'Январь', 2: 'Февраль', 3: 'Март', 4: 'Апрель',
    5: 'Май', 6: 'Июнь', 7: 'Июль', 8: 'Август',
    9: 'Сентябрь', 10: 'Октябрь', 11: 'Ноябрь', 12: 'Декабрь'
}

months_order = [
    'Январь', 'Февраль', 'Март', 'Апрель',
    'Май', 'Июнь', 'Июль', 'Август',
    'Сентябрь', 'Октябрь', 'Ноябрь', 'Декабрь'
]

traf_accident_kirov['month'] = pd.Categorical(
    traf_accident_kirov['datetime'].dt.month.map(months_ru),
    categories=months_order,
    ordered=True
)

traf_accident_kirov.head()
```

#### Результат группировки категорий аварий

```python
# Пользовательская группировка

def remap_category(cat):
    if cat == 'Столкновение':
        return 'Столкновение'
    elif cat in ('Наезд на пешехода', 'Наезд на велосипедиста'):
        return 'Наезд на пешехода'          # велосипедисты объединены с пешеходами
    elif cat == 'Съезд с дороги':
        return 'Съезд с дороги'
    elif cat == 'Опрокидывание':
        return 'Опрокидывание'
    elif cat == 'Падение пассажира':
        return 'Падение пассажира'
    elif cat in ('Наезд на препятствие', 'Наезд на стоящее ТС'):
        return 'Наезд иной'                  # препятствие + стоящее ТС → «Наезд иной»
    else:
        return 'Редкие'                      # все остальные категории

grouped_cat_dtp = traf_accident_kirov['category'].value_counts().groupby(remap_category).sum().sort_values(ascending=False)
grouped_cat_dtp
```

### Что куда попало

| Группа | Комбинации |
|---|---|
| **Ясно** | `['Ясно']`, `['Ясно', 'Температура выше +30С']`, `['Ясно', 'Температура ниже -30С']` |
| **Пасмурно** | `['Пасмурно']`, `['Пасмурно', 'Температура ниже -30С']`, `['Пасмурно', 'Ураганный ветер']` |
| **Дождь** | `['Дождь']`, `['Пасмурно', 'Дождь']`, `['Дождь', 'Туман']` |
| **Снег/Метель** | `['Снегопад']`, `['Пасмурно', 'Снегопад']`, `['Метель']`, `['Пасмурно', 'Метель']`, `['Снегопад', 'Метель']` |
| **Смешанные осадки** | `['Дождь', 'Снегопад']` |
| **Туман** | `['Туман']`, `['Пасмурно', 'Туман']`, `['Ясно', 'Туман']` |

### Логика приоритетов

1. **Смешанные осадки** проверяются первыми — одновременное наличие дождя и снега/метели выделяется в отдельную группу, так как это самая опасная комбинация для вождения (гололёд + слякоть + плохая видимость).
2. **Дождь** — любая комбинация с дождём, но без снега. `['Дождь', 'Туман']` остаётся здесь, а не в «Туман», потому что осадки важнее.
3. **Снег/Метель** — снег или метель без дождя. `['Снегопад', 'Метель']` и `['Пасмурно', 'Метель']` идут сюда же.
4. **Туман** — туман без осадков. Видимость снижена, но дорога сухая.
5. **Ясно / Пасмурно** — фоновые условия без осадков и тумана. Экстремальные температуры и ураганный ветер не образуют собственных групп, а уходят в «Ясно» или «Пасмурно» по облачности.

```python
# Пользовательская группировка для погодных условий
RAIN = {'Дождь'}
SNOW = {'Снегопад', 'Метель'}

def group_weather(cond_str):
    items = set(eval(cond_str))
    
    # 1. Дождь + снег/метель одновременно → отдельная группа
    if items & RAIN and items & SNOW:
        return 'Смешанные осадки'
    
    # 2. Только дождь
    if items & RAIN:
        return 'Дождь'
    
    # 3. Только снег / метель
    if items & SNOW:
        return 'Снег/Метель'
    
    # 4. Туман без осадков
    if 'Туман' in items:
        return 'Туман'
    
    # 5. Ясно
    if 'Ясно' in items:
        return 'Ясно'
    
    # 6. Пасмурно (включая экстремальную температуру и ураганный ветер)
    return 'Пасмурно'

grouped_cat_weather = traf_accident_kirov['weather'].value_counts().groupby(group_weather).sum().sort_values(ascending=False)
grouped_cat_weather
```

#### 4. Сезонный контекст
Сезонность сильно смешивает эффект погоды:
- Летом преобладает ясно → больше столкновений.
- Осенью и зимой преобладает пасмурно → больше наездов на пешеходов.
- Снег и метель приходятся на ноябрь–февраль → в эти месяцы повышена доля съездов и опрокидываний.

**Важно:** наблюдаемая связь «погода → категория ДТП» частично объясняется сезоном, а не самой погодой. Чтобы отделить эффект погоды от сезонного, нужен дополнительный анализ внутри одного месяца или сезона, а также учёт освещённости и состояния покрытия.

#### 5. Ограничения анализа
1. Погода указана в момент ДТП и не обязательно является его причиной.
2. Группа «Туман» и «Смешанные осадки» малочисленны — выводы по ним статистически не точны.
3. Кировская область — один регион; можно проверить закономерности на Московской области.

#### 6. Практические выводы
- **Летом в ясную погоду** основная проблема — столкновения. Меры: контроль скорости, разметка, камеры на аварийных перекрёстках.
- **Осенью и зимой в пасмурную погоду** растёт число наездов на пешеходов. Меры: освещение переходов, световозвращающие элементы, снижение скорости в жилых зонах.
- **В снегопад и метель** повышена доля съездов с дороги и опрокидываний. Меры: своевременная обработка покрытия, предупреждения о гололёде, ограничение скорости на загородных трассах.
- **Туман и смешанные осадки** требуют отдельного мониторинга — данных пока мало, но эти условия потенциально самые опасные.

```python
# Пользовательская группировка по покрытию

ICE = {'Гололедица', 'Низкие сцепные качества покрытия'}
SNOW = {'Заснеженное', 'Со снежным накатом', 'Недостатки зимнего содержания'}
TREATED = {'Обработанное противогололедными материалами'}

def group_surface(cond_str):
    items = set(eval(cond_str))
    
    # 1. Гололедица / низкие сцепные качества — высший приоритет
    if items & ICE:
        return 'Гололедица'
    
    # 2. Снег, накат, недостатки зимнего содержания
    if items & SNOW:
        return 'Снег / накат'
    
    # 3. Обработанное противогололедными (без др. зимних)
    if items & TREATED:
        return 'Обработанное'
    
    # 4. Мокрое
    if 'Мокрое' in items:
        return 'Мокрое'
    
    # 5. Сухое (вкл. комбинации с дефектами разметки, знаков, освещения)
    if 'Сухое' in items:
        return 'Сухое'
    
    # На случай непредвиденных комбинаций
    return 'Прочее'

grouped_sur = traf_accident_kirov['road_conditions'].value_counts().groupby(group_surface).sum().sort_values(ascending=False)
grouped_sur
```

#### 3.1. Столкновения — светлое время суток, сухое покрытие
- Лидирующие комбинации по всей выборке:

  - **июнь, ясно, светло, сухое — 316**;
  - **август, ясно, светло, сухое — 287**;
  - **май, ясно, светло, сухое — 206**.
- Столкновения — это «дневная сухая» категория. Основной вклад дают летние месяцы в ясную погоду.
- Практический вывод: профилактика столкновений должна работать **днём**, а не только в плохую погоду. Контроль скорости, разметка, камеры на аварийных перекрёстках эффективны именно в этих условиях.

#### 3.2. Наезды на пешеходов — тёмное время суток + мокрое покрытие
- Яркие пики:
  - **октябрь, пасмурно, темно (освещение включено), мокрое — 91**;

- Практический вывод: наибольший риск наезда на пешехода — поздней осенью вечером. Меры: освещение переходов, световозвращающие элементы, снижение скорости в жилых зонах, «умные» переходы.

#### 3.3. Съезды с дороги и опрокидывания — снег/накат, гололёд

- В январе заметны комбинации:

- Практический вывод: зимой снег и накат — триггер потери управления. Меры: своевременная обработка покрытия, предупреждения о гололёде, ограничение скорости на загородных трассах.

#### 3.4. Обработанное покрытие — индикатор зимних ДТП
- Отдельная категория «Обработанное противогололёдными материалами» даёт 1 479 ДТП (10.2%).

#### 3.5. Отсутствие освещения ночью — «слепая зона»
- Категории с наибольшим риском при отсутствии освещения: **наезды на пешеходов, съезды с дороги, наезды на препятствие**.

#### 4. Взаимосвязь освещённости и покрытия

- **Светлое + сухое** — 55% всех ДТП; здесь преобладают **столкновения** и **наезды на пешеходов** в тёплый сезон.

#### Ключевые различия

1. **Структура категорий:** в Кировской области выше доля наездов на пешеходов (34% против 29%), в Московской — выше доля «Наезд иной» и «Съезд с дороги». Это может быть связано с более интенсивным движением в МО.
2. **Снег и метель:** в Московской области их доля выше (7% против 5.2%) — климат холоднее, снежный период длиннее.
3. **Общее число ДТП:** в МО на порядок больше абсолютных значений, но относительные пропорции сохраняются.
