### Обработка пропущенных значений и аномалий

````python
# Удаляем строки без названия игры (всего 2)
games = games.dropna(subset=['name'])
````

````python
# year_of_release можно заполнить медианой по платформе
games['year_of_release'] = games['year_of_release'].fillna(
games.groupby('platform')['year_of_release'].transform('median'))
````

````python
# Заполняем рейтинги значением '-1' / -1
games['critic_score'] = games['critic_score'].fillna(-1)
games[['rating', 'user_score']] = (
    games[['rating', 'user_score']].fillna('-1')
)
````

````python
# Обработка аномальных значений в столбцах с выручкой
games[['eu_sales', 'jp_sales', 'other_sales']] = games[['eu_sales', 'jp_sales', 'other_sales']].replace(['', 'NA', 'N/A', 'null', '?', '-', 'unknown'], '')
````

````python
# Проверяем, что games очистился
games.isna().sum()
````

````python
# Выводим информацию по games
games.info()
````

### Корректировка типа данных

Для оптимизации возможно применение числовых типов с уменьшенной разрядностью

````python
# Преобразовываем типы данных с понижением разрядности
for col in ['na_sales',
               'eu_sales',
               'jp_sales',
               'other_sales',
               'critic_score',
               'user_score']:
    games[col] = pd.to_numeric(games[col], errors='coerce', downcast='float')
````

````python
# Преобразовываем типы данных с понижением разрядности
for col in ['year_of_release']:
    games[col] = pd.to_numeric(games[col], errors='coerce', downcast='integer')
````

## Фильтрация данных

````python
# Создаем новый датафрейм на основе среза games по годам (2000 - 2013)
df_actual = games[(games['year_of_release'] >= 2000) & (games['year_of_release'] <= 2013)].copy()
df_actual.head()
````

## Категоризация данных

````python
# Категоризация данных по оценкам критиков
def categ_critic(value):
    if value >= 0 and value < 30:
        return 'низкая оценка'
    elif value >= 30 and value < 80:
        return 'средняя оценка'
    elif value >= 80 and value <= 100:
        return 'высокая оценка'
    else:
        return 'нет оценки'

df_actual['group_critic_score'] = df_actual['critic_score'].apply(categ_critic)
````

````python
# Выводим количество игр по категориям оценки критиков
df_actual.groupby('group_critic_score')['name'].count()
````

````python
# Категоризация данных по оценкам игроков
def categ_user(value):
    if value >= 0 and value < 3:
        return 'низкая оценка'
    elif value >= 3 and value < 8:
        return 'средняя оценка'
    elif value >= 8 and value <= 10:
        return 'высокая оценка'
    else:
        return 'нет оценки'

df_actual['group_user_score'] = df_actual['user_score'].apply(categ_user)
````

````python
# Выводим количество игр по категориям оценки игроков
df_actual.groupby('group_user_score')['name'].count()
````
